```python
import pandas as pd

# Load the CSV file into a DataFrame named 'cars'
cars = pd.read_csv('cars.csv')
```

# A. POSITIONAL AND LABEL-BASED SLICING

```python
# a. Display the shape and complete list of column names
print("Shape of DataFrame:", cars.shape)
print("Column names:", list(cars.columns))

# b. Extract rows 6 through 10 using positional slicing (1-based row 1 is index 0)
cars_6_to_10 = cars.iloc[5:10]

# c. Select specific columns using column labels
cars_6_to_10_subset = cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]

# Display the subsetted DataFrame
cars_6_to_10_subset
```

# B. MODEL LOOKUP

```python
# a. Boolean lookup for Toyota Corolla (all columns)
toyota = cars.loc[cars['Model'] == 'Toyota Corolla']

# b. Boolean lookup for Pontiac Firebird (selected columns)
pontiac = cars.loc[cars['Model'] == 'Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]

# Display both DataFrames
display(toyota)
display(pontiac)
```

# C. MULTI-MODEL SUBSETTING

```python
# Target models list
target_models = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']

# Target columns list
target_columns = ['Model', 'mpg', 'cyl', 'hp', 'gear']

# Filter DataFrame by model values and retain specified columns
selected_cars = cars.loc[cars['Model'].isin(target_models), target_columns]

# Display DataFrame and shape check
display(selected_cars)
print("DataFrame Shape:", selected_cars.shape)  # Expected output: (3, 5)
```
